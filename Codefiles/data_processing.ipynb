```python
final_cols = [
    '_PATIENT',
    
    # Demographics
    'age_at_initial_pathologic_diagnosis',
    'gender',
    
    # Tumor stage (we will combine)
    'pathologic_stage',
    'AJCC_Stage_nature2012',
    
    # Subtype (main biological signal)
    'PAM50Call_RNAseq',
    
    # Receptor status
    'ER_Status_nature2012',
    'PR_Status_nature2012',
    'HER2_Final_Status_nature2012',
    
    # Survival
    'OS_Time_nature2012',
    'OS_event_nature2012'
]

# Keep only existing columns
final_cols = [col for col in final_cols if col in phenotype.columns]

phenotype_final = phenotype[final_cols]
phenotype_final
```

```python
missing_percent2 = phenotype_final.isnull().mean() * 100

# Sort to see worst columns
missing_percent2 = missing_percent2.sort_values(ascending=False)

print(missing_percent2)
```

```python
import pandas as pd

# -------------------------------
# Helper functions
# -------------------------------

def smart_mode(series):
    series = series.dropna()
    if len(series) == 0:
        return None
    return series.value_counts().idxmax()

# Stage ranking (biologically meaningful)
stage_order = {
    'Stage I': 1, 'Stage IA': 1, 'Stage IB': 1,
    'Stage II': 2, 'Stage IIA': 2, 'Stage IIB': 2,
    'Stage III': 3, 'Stage IIIA': 3, 'Stage IIIB': 3, 'Stage IIIC': 3,
    'Stage IV': 4
}

def max_stage(series):
    series = series.dropna()
    if len(series) == 0:
        return None
    
    mapped = series.map(stage_order)
    if mapped.notna().sum() == 0:
        return smart_mode(series)  # fallback
    
    return series.loc[mapped.idxmax()]

# -------------------------------
# Combine duplicate columns FIRST
# -------------------------------

# Stage
if 'pathologic_stage' in phenotype_final.columns and 'AJCC_Stage_nature2012' in phenotype_final.columns:
    phenotype_final['final_stage'] = phenotype_final['pathologic_stage'].fillna(
        phenotype_final['AJCC_Stage_nature2012']
    )
elif 'pathologic_stage' in phenotype_final.columns:
    phenotype_final['final_stage'] = phenotype_final['pathologic_stage']
else:
    phenotype_final['final_stage'] = phenotype_final['AJCC_Stage_nature2012']

# Drop old stage columns
phenotype_final = phenotype_final.drop(columns=[
    col for col in ['pathologic_stage', 'AJCC_Stage_nature2012'] if col in phenotype_final.columns
])

# -------------------------------
# Aggregate per patient
# -------------------------------

agg_dict = {}

for col in phenotype_final.columns:
    
    if col == '_PATIENT':
        continue
    
    elif col == 'final_stage':
        agg_dict[col] = max_stage
        
    elif col == 'OS_Time_nature2012':
        agg_dict[col] = 'max'
        
    elif col == 'OS_event_nature2012':
        agg_dict[col] = 'max'
        
    elif phenotype_final[col].dtype == 'object':
        agg_dict[col] = smart_mode
        
    else:
        agg_dict[col] = 'mean'

# Apply aggregation
phenotype_summary = phenotype_final.groupby('_PATIENT', as_index=False).agg(agg_dict)

# -------------------------------
# Final check
# -------------------------------

print("Final shape:", phenotype_summary.shape)
print(phenotype_summary.head())
```
